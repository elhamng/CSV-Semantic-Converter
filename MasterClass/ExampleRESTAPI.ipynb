To interact with a SPARQL endpoint via a REST API in Python, you can use the requests library to make HTTP calls. Below is an example of how to query a SPARQL endpoint using a REST API in Python. This example demonstrates how to get data from a SPARQL endpoint and visualize the results.

You are querying a SPARQL endpoint (let's say, a public endpoint from DBpedia) to fetch information about a certain resource (e.g., "Albert Einstein")

```python
# Define the SPARQL endpoint URL (DBpedia public SPARQL endpoint in this case)
sparql_endpoint = "https://dbpedia.org/sparql"

#You can replace it with the URL of your desired SPARQL endpoint. 
```

```python
# Define the SPARQL query you want to execute
sparql_query = """
PREFIX dbo: <http://dbpedia.org/ontology/>
PREFIX dbr: <http://dbpedia.org/resource/>
SELECT ?person ?birthPlace WHERE {
  ?person dbo:birthPlace ?birthPlace .
  ?person rdfs:label "Albert Einstein"@en .
}
"""

# Set up the parameters for the REST API request (SPARQL query and format)
params = {
    'query': sparql_query,
    'format': 'application/json'  # Request the response in JSON format
}

```

```python
# Send the GET request to the SPARQL endpoint
response = requests.get(sparql_endpoint, params=params)
```

```python

# Check if the request was successful
if response.status_code == 200:
    # Parse the JSON response
    data = response.json()

    # Extract and display the results
    print("Results:")
    for result in data['results']['bindings']:
        person = result['person']['value']
        birth_place = result['birthPlace']['value']
        print(f"Person: {person}, Birth Place: {birth_place}")
else:
    print(f"Error: {response.status_code} - {response.text}")
```

## Make a Get request

The example here finds the latitude, longitude, and formatted address of a given location by sending a GET request to the Google Maps API. An API (Application Programming Interface) allows you to access the internals of a program in a limited way. And in most cases, the data provided is in JSON (JavaScript Object Notation) format (which is implemented as dictionary objects in Python!).

```python
# importing the requests library
import requests

# api-endpoint
URL = "http://maps.googleapis.com/maps/api/geocode/json"

# location given here
location = "delhi technological university"

# defining a params dict for the parameters to be sent to the API
PARAMS = {'address':location}

# sending get request and saving the response as response object
r = requests.get(url = URL, params = PARAMS)

# extracting data in json format
data = r.json()

```

```python
data
```

```python

# extracting latitude, longitude and formatted address
# of the first matching location
latitude = data['results'][0]['geometry']['location']['lat']
longitude = data['results'][0]['geometry']['location']['lng']
formatted_address = data['results'][0]['formatted_address']

# printing the output
print("Latitude:%s\nLongitude:%s\nFormatted Address:%s"
    %(latitude, longitude,formatted_address))
```

 Below is an example of creating a REST API with Python to interact with a SPARQL endpoint, query RDF data, and serve it as JSON. This simulates the concept of a "data station" for SPARQL.
 
Data source: SPARQL endpoint of a dataset (e.g., DBpedia or Wikidata).

Goal: Query RDF data (e.g., find information about programming languages) and provide it via a REST API.

Tool: Python with Flask and SPARQLWrapper librari

```python
from flask import Flask, request, jsonify
from SPARQLWrapper import SPARQLWrapper, JSON
```

```python
# Initialize Flask app
app = Flask(__name__)
```

```python
# Define the SPARQL endpoint (e.g., Wikidata or DBpedia)
SPARQL_ENDPOINT = "https://dbpedia.org/sparql"
```

```python

@app.route('/sparql-query', methods=['GET'])
def sparql_query():
    # Get query parameters
    query = request.args.get('query')
    
    if not query:
        return jsonify({"error": "Please provide a SPARQL query as the 'query' parameter."}), 400

    try:
        # Initialize SPARQL wrapper
        sparql = SPARQLWrapper(SPARQL_ENDPOINT)
        sparql.setQuery(query)
        sparql.setReturnFormat(JSON)
        
        # Execute query
        results = sparql.query().convert()
        return jsonify(results)  # Return SPARQL results as JSON
    except Exception as e:
        return jsonify({"error": str(e)}), 500


```

```python
# Example route for specific queries (optional)
@app.route('/programming-languages', methods=['GET'])
def get_programming_languages():
    query = """
    PREFIX dbo: <http://dbpedia.org/ontology/>
    SELECT ?language ?label WHERE {
      ?language a dbo:ProgrammingLanguage .
      ?language rdfs:label ?label .
      FILTER (lang(?label) = 'en')
    } LIMIT 10
    """
    try:
        # Initialize SPARQL wrapper
        sparql = SPARQLWrapper(SPARQL_ENDPOINT)
        sparql.setQuery(query)
        sparql.setReturnFormat(JSON)
        
        # Execute query
        results = sparql.query().convert()
        return jsonify(results)
    except Exception as e:
        return jsonify({"error": str(e)}), 500

```

```python
if __name__ == "__main__":
    app.run(debug=True)
```
