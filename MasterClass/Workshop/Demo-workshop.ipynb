````python
#assignment: Check if a number is even or odd
user_input = input ("inter a number: ")
try:
    number = int(user_input)
    
    if number % 2 == 0:
        print("That is an even number")
    else:
        print("That is an odd number")
except ValueError:
    print("That is not an integer, please enter a valid integer.")
````

````python
import rdflib
from rdflib import Graph, Literal, RDF, URIRef, Namespace
from rdflib.namespace import RDFS, XSD
````

````python
default_namespace = "http://example.org#"  # Replace with your desired namespace

# Define common namespace URIs
#version = 3.0.0

onz_one_s = 'http://purl.org/ozo/onz-org#'
onz_one_ns = Namespace(onz_org_s)
onz_two_s = 'http://purl.org/ozo/onz-g#'
onz_two_ns = Namespace(onz_two_s)
onz_three_s = 'http://purl.org/ozo/onz-three#'
onz_three_ns = Namespace(onz_three_s)
````

````python
g = rdflib.Graph()
````

````python
# Bind the namespace as the default namespace
g.bind("", default_namespace)
g.bind('onz-one', onz_one_ns)
g.bind('onz-two', onz_two_ns)
g.bind('onz-three', onz_three_ns)
````

````python
headers = df.columns.tolist()
for index, row in df.iterrows():
    try:
        # Validate required fields exist
        if pd.isna(row['ContractID']):
            print(f"Row {index} skipped: 'ContractID' is missing.")
            continue

        # Direct column access and replace spaces with underscores
        contract_id = str(row['ContractID']).strip().replace(' ', '_')           
        contract_uri = URIRef(f"{default_namespace}Agreement_{contract_id}")

        employee_id = str(row['PersonID']).strip().replace(' ', '_') if pd.notna(row['PersonID']) else "Unknown"
        employee_uri = URIRef(f"{default_namespace}Employee_{employee_id}")
        g.add((employee_uri, RDFS.label, Literal(employee_id, datatype=XSD.string)))


        date_start = row['DateStart'] if pd.notna(row['DateStart']) else None
        date_end = row['DateEnd'] if pd.notna(row['DateEnd']) else None
        g.add((contract_uri, onz_two_ns.startDatum, Literal(date_start, datatype=XSD.date)))
        if date_end:
            g.add((contract_uri, onz_two_ns.eindDatum, Literal(date_end, datatype=XSD.date)))

        
        # Add to RDF graph
        g.add((contract_uri, RDF.type, onz_three_ns.Type))
        g.add((contract_uri, onz_three_ns.employee, employee_uri))
        g.add((employee_uri, RDF.type, onz_three_ns.Human))
        
        print(f"Processed row {index} successfully.")
       
    except Exception as e:
        print(f"Error processing row {index}: {e}")
````

````python
# Save the RDF graph to a TTL file
output_file = "contract_data.ttl"
g.serialize(destination=output_file, format='turtle')
print(f"RDF graph saved to {output_file}")

# Optional: Print the TTL content to see what was generated
print("\nGenerated TTL content:")
print(g.serialize(format='turtle'))
````
